<h2 align="center">Statistical Analysis</h2>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import (
    pearsonr,
    spearmanr,
    shapiro,
    levene,
    f_oneway,
    kruskal
)
import statsmodels.api as sm
import statsmodels.formula.api as smf
from statsmodels.stats.outliers_influence import variance_inflation_factor
from pathlib import Path
```

```python
df = pd.read_excel('../../data/processed/3rd_one_year_mine_data_cleaned.xlsx')
df.columns = df.columns.str.strip()
df.head()
```

```python
df = pd.get_dummies(df, columns=['Wind_direction'])
```

```python
df.columns
```

<h2 align="center">Shapiro-Wilk test for normality</h2>
<p align="center">Tests whether a dataset is drawn from a normal distribution</p>

```python
for column in [
    "Temperature",
    "Relative_humidity",
    "RF-CUMULATIVE",
    "Wind_speed",
    'Wind_direction_NorthWest',
    "Distance",
    'Production_ROM',
    'Production_OB',
    'Production_ROM+OB',
    "Vehicle Capacity",
    "PM10",
    "PM2.5",
]:

    stat, p = shapiro(df[column].sample(500, random_state=42))

    print(column)

    print(f"P-value : {p:.4f}")

    print()
```

<h2 align="center">Pearson correlation coefficient</h2>
<p align="center">The coefficient measures the linear relationship between two datasets, ranging from -1 (perfect negative correlation) to +1 (perfect positive correlation), with 0 indicating no linear correlation.</p>

```python
weather = [
    "Temperature",
    "Relative_humidity",
    "RF-CUMULATIVE",
    "Wind_speed",
    'Wind_direction_NorthWest',
]

for feature in weather:

    r, p = pearsonr(df[feature], df["PM10"])

    print(f"{feature}")

    print(f"Correlation : {r:.3f}")

    print(f"P-value : {p:.4f}")

    print()
```

```python
mine = [
    "Distance",
    'Production_ROM',
    'Production_OB',
    'Production_ROM+OB',
    "Vehicle Capacity",
]

for feature in mine:

    r, p = pearsonr(df[feature], df["PM10"])

    print(f"{feature}")

    print(f"Correlation : {r:.3f}")

    print(f"P-value : {p:.4f}")

    print()
```

<h2 align="center">Spearman rank-order correlation coefficient</h2>
<p align="center">This nonparametric measure evaluates the monotonic relationship between two datasets, with values ranging from -1 (perfect negative correlation) to +1 (perfect positive correlation).</p>

```python
weather = [
    "Temperature",
    "Relative_humidity",
    "RF-CUMULATIVE",
    "Wind_speed",
    'Wind_direction_NorthWest',
]

for feature in weather:

    r, p = spearmanr(df[feature], df["PM10"])

    print(f"{feature}")

    print(f"Correlation : {r:.3f}")

    print(f"P-value : {p:.4f}")

    print()
```

```python
mine = [
    "Distance",
    'Production_ROM',
    'Production_OB',
    'Production_ROM+OB',
    "Vehicle Capacity",
]

for feature in mine:

    r, p = spearmanr(df[feature], df["PM10"])

    print(f"{feature}")

    print(f"Correlation : {r:.3f}")

    print(f"P-value : {p:.4f}")

    print()
```

```python
X = df[
    [
        "Temperature",
        "Relative_humidity",
        "RF-CUMULATIVE",
        "Wind_speed",
        'Wind_direction_NorthWest',
        "Distance",
        'Production_ROM',
        'Production_OB',
        'Production_ROM+OB',
        "Vehicle Capacity",
    ]
]
y = df["PM10"]
X = sm.add_constant(X)
```

```python
# drop rows with NaNs/inf in X or y before fitting
valid_idx = y.notna() & X.notna().all(axis=1)
model = sm.OLS(y.loc[valid_idx], X.loc[valid_idx].astype(float)).fit()
```

```python
print(model.summary())
```
