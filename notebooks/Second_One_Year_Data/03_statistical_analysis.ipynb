<h2 align="center">Shapiro-Wilk test for normality</h2>
<p align="center">Tests whether a dataset is drawn from a normal distribution</p>

```python
for column in [
    "Temperature",
    "Relative_humidity",
    "RF-CUMULATIVE",
    "Wind_speed",
    "Road_width",
    "Distance between the dust generation/source point and the haul road monitoring station",
    "Production_ROM",
    "Production_OB",
    "Production_ROM+OB",
    "Vehicle capacity",
    "LOCATION",
    "SEASON",
    "PM10",
    "PM2.5"
]:

    if not pd.api.types.is_numeric_dtype(df[column]):
        continue

    sample = df[column].dropna()
    if len(sample) > 500:
        sample = sample.sample(500, random_state=42)

    stat, p = shapiro(sample)

    print(column)

    print(f"P-value : {p:.4f}")

    print()
```

<h2 align="center">Pearson correlation coefficient</h2>
<p align="center">The coefficient measures the linear relationship between two datasets, ranging from -1 (perfect negative correlation) to +1 (perfect positive correlation), with 0 indicating no linear correlation.</p>

```python
weather = [
    "Temperature",
    "Relative_humidity",
    "RF-CUMULATIVE",
    "Wind_speed",
    "Road_width",
    "Distance between the dust generation/source point and the haul road monitoring station",
    "Production_ROM",
    "Production_OB",
    "Production_ROM+OB",
    "Vehicle capacity",
    "LOCATION",
    "SEASON",
    "PM10",
    "PM2.5"
]

target = "PM10"

for feature in weather:
    x = pd.to_numeric(df[feature], errors="coerce")
    y = pd.to_numeric(df[target], errors="coerce")

    valid = x.notna() & y.notna()
    x_valid = x.loc[valid]
    y_valid = y.loc[valid]

    print(f"{feature}")

    if x_valid.nunique() < 2 or y_valid.nunique() < 2:
        print("Correlation : not enough numeric variation")
        print("P-value : nan")
        print()
        continue

    r, p = pearsonr(x_valid, y_valid)

    print(f"Correlation : {r:.3f}")

    print(f"P-value : {p:.4f}")

    print()
```

<h2 align="center">Spearman rank-order correlation coefficient</h2>
<p align="center">This nonparametric measure evaluates the monotonic relationship between two datasets, with values ranging from -1 (perfect negative correlation) to +1 (perfect positive correlation).</p>

```python
weather = [
    "Temperature",
    "Relative_humidity",
    "RF-CUMULATIVE",
    "Wind_speed",
    "Road_width",
    "Distance between the dust generation/source point and the haul road monitoring station",
    "Production_ROM",
    "Production_OB",
    "Production_ROM+OB",
    "Vehicle capacity",
    "LOCATION",
    "SEASON",
    "PM10",
    "PM2.5"
]

for feature in weather:

    r, p = spearmanr(df[feature], df["PM10"])

    print(f"{feature}")

    print(f"Correlation : {r:.3f}")

    print(f"P-value : {p:.4f}")

    print()
```
