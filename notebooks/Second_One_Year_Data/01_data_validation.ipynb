<h2 align="center">Importing Required Libraries</h2>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
import missingno as msno
from pathlib import Path
```

<h2 align="center">Data Loading</h2>

```python
try:
    df = pd.read_excel(
        '../../data/raw/Second_One_Year_Data.xlsx',
        header=[0, 1]
        )
    print("Data Frame Head : ")
    print(df.head())
    print("Data Frame Tail : ")
    print(df.tail())
    print("Data Frames Samples : ")
    print(df.sample())
    print("Data Frame Information : ")
    print(df.info())
    print("Data Frame Description : ")
    print(df.describe())

except FileNotFoundError:
    print("Error: The CSV file not found.")
except pd.errors.EmptyDataError:
    print("Error: The CSV file is empty.")
except pd.errors.ParserError:
    print("Error: The CSV file is malformed.")
except Exception as e:
    print(f"Unexpected error occurred: {e}")
```

<h2 align="center">Flattening columns</h2>

```python
df.columns = [
    "_".join(
        [str(i) for i in col if "Unnamed" not in str(i)]
    ).strip("_")
    for col in df.columns
]
```

<h2 align="center">Removing Unneseasry columns</h2>

```python
# Removing NaN columns
df = df.dropna(axis=1, how="all")

# Drop columns nammed "Unnamed"
df = df.loc[:, ~df.columns.str.contains(r"^Unnamed")]

# Remove dupicates
df = df.loc[:, ~df.columns.duplicated()]

print(df.columns)
```

<h2 align="center">Naming Inconsistensies</h2>

```python
df.columns = df.columns.str.strip()
```
