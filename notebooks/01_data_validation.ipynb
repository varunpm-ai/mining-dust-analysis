<h2 align="center">Importing Required Libraries</h2>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
import missingno as msno
from pathlib import Path
```

<h2 align="center">Data Loading</h2>

```python
try:
    df = pd.read_excel('../data/raw/DATA SHEET.xlsx')
    print("Data Frame Head : ")
    print(df.head())
    print("Data Frame Tail : ")
    print(df.tail())
    print("Data Frames Samples : ")
    print(df.sample())
    print("Data Frame Information : ")
    print(df.info())
    print("Data Frame Description : ")
    print(df.describe())

except FileNotFoundError:
    print("Error: The CSV file not found.")
except pd.errors.EmptyDataError:
    print("Error: The CSV file is empty.")
except pd.errors.ParserError:
    print("Error: The CSV file is malformed.")
except Exception as e:
    print(f"Unexpected error occurred: {e}")
```

<h2 align="center">Removing Unneseasry columns</h2>

```python
# Removing Nan columns

df = df.dropna(axis=1, how="all")
print(df.columns)
```

<h2 align="center">Renaming Column names </h2>

```python
column_mapping = {
    "T": "Temperature",
    "RH": "Relative_humidity",
    "RF-CUMULATIVE": "Cumulative_rainfall",
    "WS": "Wind_speed",
}

df = df.rename(columns=column_mapping)
```

<h2 align="center">Categorical validation</h2>

```python
categorical_columns = ["SEASON", "MONTH", "LOCATION"]

for column in categorical_columns:
    print("=" * 60)
    print(f"{column}")
    print("=" * 60)

    print(f"\nNumber of Unique Values : {df[column].nunique()}\n")

    print(df[column].value_counts(dropna=False))

    print("\nUnique Values:")
    print(sorted(df[column].unique()))

    print("\n")
```
