```python
df.columns = df.columns.str.strip()

weather_features = [
    "Temperature",
    "Relative_humidity",
    "RF-CUMULATIVE",
    "Wind_speed"
]

targets = ["PM10", "PM2.5"]
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

axes = axes.flatten()

for i, feature in enumerate(weather_features):

    sns.regplot(
        data=df,
        x=feature,
        y="PM10",
        scatter_kws={"alpha":0.5},
        line_kws={"color":"red"},
        ax=axes[i]
    )

    axes[i].set_title(f"{feature} vs PM10")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

axes = axes.flatten()

for i, feature in enumerate(weather_features):

    sns.regplot(
        data=df,
        x=feature,
        y="PM2.5",
        scatter_kws={"alpha":0.5},
        line_kws={"color":"red"},
        ax=axes[i]
    )

    axes[i].set_title(f"{feature} vs PM2.5")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,8))

corr = df.corr(method="spearman", numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    center=0,
    fmt=".2f"
)

plt.title("Spearman Correlation Matrix")

plt.show()
```
